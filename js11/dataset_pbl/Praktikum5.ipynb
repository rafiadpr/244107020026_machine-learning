# Praktikum 5 : threshold dan SVM pada teks

```python
# Langkah 1 - Memuat dan memisahkan teks
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
data = pd.read_csv('../../dataset_pbl/reviews.csv')
data = data[data['rating'].isin([1,5])].head(1000).copy()
data['text'] = data['text'].fillna('').str.split().str.join(' ').str.lower()
y = data['rating'].map({1:0,5:1}).to_numpy()
text_train, text_test, y_train, y_test = train_test_split(data['text'],y,test_size=0.2,random_state=42,stratify=y)
```

```python
# Langkah 2 - Inspeksi
data[['text','rating']].head()
```

```python
# Langkah 3–4 - Fitur jumlah kata menggantikan rata-rata kecerahan
X_train = text_train.str.split().str.len().to_numpy().reshape(-1,1)
X_test = text_test.str.split().str.len().to_numpy().reshape(-1,1)
print('Ukuran fitur latih:', X_train.shape)
print('Ukuran fitur uji:', X_test.shape)
```

```python
# Langkah 5 - Klasifikasi threshold
def predict_label(jumlah_kata, threshold):
    return int(jumlah_kata>threshold)
pred_train = np.array([predict_label(v,20) for v in X_train[:,0]])
pred_test = np.array([predict_label(v,20) for v in X_test[:,0]])
```

```python
# Langkah 6 - Evaluasi threshold
from sklearn.metrics import accuracy_score
print('Akurasi threshold train:',accuracy_score(y_train,pred_train))
print('Akurasi threshold test:',accuracy_score(y_test,pred_test))
```

```python
# Alternatif SVM dengan satu fitur yang sama
from sklearn.svm import SVC
model = SVC()
model.fit(X_train,y_train)
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)
```

```python
print('Akurasi SVM train:',accuracy_score(y_train,y_train_pred))
print('Akurasi SVM test:',accuracy_score(y_test,y_test_pred))
```
