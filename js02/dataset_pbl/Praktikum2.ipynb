# Praktikum 2: imputasi dataset PBL

```python
## Langkah 0 - Persiapan
import pandas as pd
import numpy as np
```

```python
## Langkah 1 - Memuat Data
df = pd.read_csv('../../dataset_pbl/cafes.csv')

# Tampilkan 5 data teratas
df.head()
```

```python
## Langkah 2 - Inspeksi Data
df.shape
df.info()
df.isnull().sum()
df.describe()
```

```python
## Langkah 3 - Data Imputation
# price_min dan price_max - mean
df['price_min'] = df['price_min'].fillna(df['price_min'].mean())
df['price_max'] = df['price_max'].fillna(df['price_max'].mean())

# price_range - teks pengganti
df['price_range'] = df['price_range'].fillna("Tidak tersedia")

# categories - modus
df['categories'] = df['categories'].fillna(df['categories'].mode()[0])
```

```python
## Langkah 4 - Validasi Data
df.info()

df.to_csv('cafes-fixed.csv', index=False)
```
