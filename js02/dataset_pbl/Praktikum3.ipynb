# Praktikum 3: encoding dan standardisasi dataset PBL

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
## Langkah 1 - Load Data
dpath = 'cafes-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

```python
## Langkah 2 - Slice Data
df = df[['rating', 'review_count', 'price_min', 'categories', 'price_range']]
df.head()
```

```python
## Langkah 3 - Encoding
le = LabelEncoder() # membuat objek dari LabelEncoder
df['categories'] = le.fit_transform(df['categories']) # proses encoding
df['price_range'] = le.fit_transform(df['price_range']) # proses encoding
```

```python
## Langkah 4 - Verifikasi Hasil
df.head()
```

```python
## Langkah 5 - Standarisasi
std = StandardScaler()
df['price_min'] = std.fit_transform(df[['price_min']])
```
