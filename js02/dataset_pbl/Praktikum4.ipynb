# Praktikum 4 : preprocess teks ulasan

```python
## Langkah 0 - Persiapan
import pandas as pd
```

```python
## Langkah 1 - Load Teks
df = pd.read_csv('../../dataset_pbl/reviews.csv')
df[['text']].head()
```

```python
## Langkah 2 - Merapikan Spasi
df['text'] = df['text'].fillna('').str.split().str.join(' ')
df[['text']].head()
```

```python
## Langkah 3 - Mengubah ke Huruf Kecil
df['text'] = df['text'].str.lower()
df[['text']].head()
```
