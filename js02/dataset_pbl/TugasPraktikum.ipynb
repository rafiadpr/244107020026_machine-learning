# Tugas praktikum: prapengolahan dataset PBL

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
## Load Dataset
df = pd.read_csv('cafes-fixed.csv')
df.head(10)
```

```python
## Membuang variabel yang tidak digunakan
df_cleaned = df.drop(columns=[
    'cafe_id', 'name', 'address', 'phone', 'website', 'price_range',
    'opening_hours', 'scraped_at', 'source_query'
])
df_cleaned.head(10)
```

```python
le = LabelEncoder()
df_cleaned['categories'] = le.fit_transform(df_cleaned['categories'])
df_cleaned.head(10)
```

```python
fitur_numerik = df_cleaned.drop(columns=['categories']).columns

scaler = StandardScaler()
df_cleaned[fitur_numerik] = scaler.fit_transform(df_cleaned[fitur_numerik])

df_cleaned.head(10)
```
