# Praktikum 1 : ekstraksi fitur data kafe

```python
## Langkah 1 - Load Data
import pandas as pd

data = '../../dataset_pbl/cafes.csv' # path dataset
df = pd.read_csv(data) # load dataset

df.head(10)
```

```python
## Langkah 2 - Pengecekan Data
df.info()
```

```python
df.isnull().sum()
```

```python
## Langkah 3 - Imputasi
# Harga - mean
df['price_min'] = df['price_min'].fillna(df['price_min'].mean())
df['price_max'] = df['price_max'].fillna(df['price_max'].mean())

# source_query - teks pengganti
df['source_query'] = df['source_query'].fillna('Tidak tersedia')

# categories - modus
df['categories'] = df['categories'].fillna(df['categories'].mode()[0])
```

```python
## Langkah 4 - Validasi Hasil
df.isnull().sum()
```

```python
df.to_csv('cafes-fixed.csv', index=False)
df.head(10)
```

```python
## Langkah 5 - Ekstraksi Fitur
import matplotlib.pyplot as plt
import seaborn as sns

# Perhitungan numerik, ekstraksi bagian teks, dan pengelompokan seperti modul
df['rentang_harga'] = df['price_max'] - df['price_min']
df['kategori_utama'] = df['categories'].str.split(',').str[0].str.strip("[]' ")
df['kelompok_harga'] = pd.cut(df['price_min'], bins=[0, 25000, 50000, 100000, float('inf')],
    labels=['Sampai 25.000', '25.000-50.000', '50.000-100.000', 'Di atas 100.000'])

df[['name', 'rentang_harga', 'kategori_utama', 'price_min', 'kelompok_harga']].head()
```

```python
## Langkah 6 - Visualisasi Hasil Ekstraksi Fitur
plt.figure(figsize=(12, 6))
sns.barplot(x='rentang_harga', y='rating', data=df)
plt.title('Rating berdasarkan rentang harga')
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x='kategori_utama', y='rating', data=df,
    order=df.groupby('kategori_utama')['rating'].mean().sort_values().index)
plt.title('Rating berdasarkan kategori utama')
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x='kelompok_harga', y='rating', data=df,
    order=['Sampai 25.000', '25.000-50.000', '50.000-100.000', 'Di atas 100.000'])
plt.title('Rating berdasarkan kelompok harga')
plt.show()
```
