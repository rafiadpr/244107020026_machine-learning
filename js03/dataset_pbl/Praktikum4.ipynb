# Praktikum 4 : ekstraksi fitur teks ulasan

```python
# Langkah 0 - Load Library
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Langkah 1 - Load Teks
df = pd.read_csv('../../dataset_pbl/reviews.csv')
df[['text']].head()
```

```python
# Langkah 2 - Ekstrak Fitur Frekuensi Kata
vectorizer = CountVectorizer()
fitur = vectorizer.fit_transform(df['text'].fillna(''))
print('Dimensi fitur:', fitur.shape)
print('Sepuluh kata pertama:', vectorizer.get_feature_names_out()[:10])
print(fitur[0])
```
