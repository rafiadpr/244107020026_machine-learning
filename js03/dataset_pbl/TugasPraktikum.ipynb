```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# 1. Memuat Data
df = pd.read_csv('cafes-fixed.csv')
```

```python
# 2. Memisahkan variabel yang tidak digunakan
df_cleaned = df.drop(columns=[
    'cafe_id', 'name', 'address', 'categories', 'phone', 'website',
    'price_range', 'opening_hours', 'scraped_at'
])
print(df_cleaned.head())
```

```python
# 3. Melakukan proses encoding pada kolom "source_query"
le = LabelEncoder()
df_cleaned['source_query'] = le.fit_transform(df_cleaned['source_query'])

print(df_cleaned[['source_query', 'price_min', 'price_max']].head())
```

```python
# 4. Melakukan proses standarisasi pada semua kolom numerik
fitur_numerik = df_cleaned.columns.drop('source_query')
scaler = StandardScaler()

df_cleaned[fitur_numerik] = scaler.fit_transform(df_cleaned[fitur_numerik])
```

```python
# 5. Melakukan proses stratified split data (Rasio 80:20)
X = df_cleaned.drop(columns=['source_query']) 
y = df_cleaned['source_query']                

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print(f"Dimensi X_train: {X_train.shape}")
print(f"Dimensi X_test: {X_test.shape}")
print("\nContoh X_train (5 baris pertama):")
print(X_train.head())
```
