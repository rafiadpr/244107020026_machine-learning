# Praktikum 2 : seleksi fitur data kafe

```python
## Langkah 0 - Load Library
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
## Langkah 1 - Load Data
df = pd.read_csv('cafes-fixed.csv')
df.head()
```

```python
## Langkah 2 - Encoding categories dan verifikasi
le = LabelEncoder()
df['categories'] = le.fit_transform(df['categories'])
print(df[['categories']].head())
```

```python
## Langkah 4 - Encoding price_range dan verifikasi
df['price_range'] = df['price_range'].fillna('Unknown')
print(df[['price_range']].head())
df['price_range'] = le.fit_transform(df['price_range'])
print(df[['price_range']].head())
```

```python
## Langkah 5 - Imputasi price_min dan verifikasi
df['price_min'] = df['price_min'].fillna(df['price_min'].median())
print(df[['price_min']].head(10))
```

```python
## Langkah 7 - Standarisasi price_min dan verifikasi
scaler = StandardScaler()
df['price_min'] = scaler.fit_transform(df[['price_min']])
print(df[['price_min']].head(10))
```

```python
## Langkah 8 - Verifikasi hasil preprocessing
df.head()
```

```python
## Langkah 9 - Seleksi Fitur
df = df[['source_query', 'review_count', 'price_min', 'categories', 'price_range']]
df.to_csv('cafes-selected.csv', index=False)
df.head()
```
