```python
# Import library
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
```

```python
# Langkah 1 - Teks ulasan menggantikan MNIST
data = pd.read_csv('../../dataset_pbl/reviews.csv')
teks = data['text'].fillna('').to_numpy()
y = data['rating'].to_numpy(dtype=int)-1
text_train, text_test, y_train, y_test = train_test_split(teks,y,test_size=0.2,random_state=42)
vectorizer = CountVectorizer(max_features=64)
X_train = vectorizer.fit_transform(text_train).toarray().astype('float32')
X_test = vectorizer.transform(text_test).toarray().astype('float32')
nilai_maks = X_train.max()
X_train = X_train/nilai_maks
X_test = X_test/nilai_maks
y_train = to_categorical(y_train,5)
y_test = to_categorical(y_test,5)
```

```python
import time
```

```python
# 2. Bangun model JST
model = Sequential([
    # Input berupa vektor frekuensi kata
    Flatten(input_shape=(64,)),
    # Hidden layer 1
    Dense(128, activation='relu'),
    # Hidden layer 2
    Dense(64, activation='relu'),
    # Output layer (5 rating)
    Dense(5, activation='softmax')
])
```

```python
# 3. Kompilasi model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
# 4. Latih model
mulai = time.perf_counter()
model.fit(X_train, y_train, epochs=10, batch_size=32)
waktu_latih = time.perf_counter() - mulai
```

```python
# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi pada data uji: {acc:.4f}")
```

```python
hasil = [('128-64, ReLU', loss, acc, waktu_latih)]
```
