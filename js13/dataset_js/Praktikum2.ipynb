## Langkah 1. Import library

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
import tensorflow as tf
```

## Langkah 2. Load dataset, one-hot encoding, dan pembagian data

```python
# Load dataset
iris = load_iris()
X = iris.data
y = iris.target.reshape(-1, 1)

# One-hot encoding
encoder = OneHotEncoder(sparse=False)
y = encoder.fit_transform(y)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

## Langkah 3. Bangun model

```python
# Bangun model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

## Langkah 4. Kompilasi dan latih model

```python
# Kompilasi
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model
model.fit(X_train, y_train, epochs=50, batch_size=8)
```

## Langkah 5. Evaluasi hasil

```python
# Evaluasi
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi: {acc}")
```

```python
hasil = [('10-8, ReLU', loss, acc)]
```

## Tugas 2

Ubah jumlah neuron hidden layer menjadi 20 dan 16.

```python
# Bangun model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(20, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

```python
# Kompilasi
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model
model.fit(X_train, y_train, epochs=50, batch_size=8)
```

```python
# Evaluasi
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi: {acc}")
hasil.append(('20-16, ReLU', loss, acc))
```

## Tugas 3

Bandingkan Sigmoid dengan ReLU pada konfigurasi 10 dan 8 neuron.

```python
# Bangun model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='sigmoid', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='sigmoid'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

```python
# Kompilasi
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model
model.fit(X_train, y_train, epochs=50, batch_size=8)
```

```python
# Evaluasi
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi: {acc}")
hasil.append(('10-8, Sigmoid', loss, acc))
```

```python
print(f"{'Konfigurasi':<20} {'Loss uji':>12} {'Akurasi uji':>12}")
for konfigurasi, loss_uji, akurasi_uji in hasil:
    print(f"{konfigurasi:<20} {loss_uji:>12.6f} {akurasi_uji:>12.4f}")
```

Semua konfigurasi menggunakan pembagian data yang sama. Hasil dapat berubah saat dijalankan ulang karena pembagian data dan bobot awal acak.
