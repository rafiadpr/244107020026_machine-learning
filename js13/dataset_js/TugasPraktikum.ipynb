# Tugas praktikum

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
```

## Langkah 1. Load dataset dan prapemrosesan

```python
# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()
# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
import time
```

## Langkah 2. Bangun model JST

```python
# 2. Bangun model JST
model = Sequential([
    # Ubah gambar 28x28 menjadi vektor
    Flatten(input_shape=(28, 28)),
    # Hidden layer 1
    Dense(128, activation='relu'),
    # Hidden layer 2
    Dense(64, activation='relu'),
    # Output layer (10 kelas)
    Dense(10, activation='softmax')
])
```

## Langkah 3. Kompilasi model

```python
# 3. Kompilasi model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

## Langkah 4. Latih model

```python
# 4. Latih model
mulai = time.perf_counter()
model.fit(X_train, y_train, epochs=10, batch_size=32)
waktu_latih = time.perf_counter() - mulai
```

## Langkah 5. Evaluasi model

```python
# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi pada data uji: {acc:.4f}")
```

```python
hasil = [('128-64, ReLU', loss, acc, waktu_latih)]
```

## Eksperimen 1. Ubah jumlah neuron menjadi 256 dan 128

```python
# 2. Bangun model JST
model = Sequential([
    # Ubah gambar 28x28 menjadi vektor
    Flatten(input_shape=(28, 28)),
    # Hidden layer 1
    Dense(256, activation='relu'),
    # Hidden layer 2
    Dense(128, activation='relu'),
    # Output layer (10 kelas)
    Dense(10, activation='softmax')
])
```

```python
# 3. Kompilasi model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
# 4. Latih model
mulai = time.perf_counter()
model.fit(X_train, y_train, epochs=10, batch_size=32)
waktu_latih = time.perf_counter() - mulai
```

```python
# 2. Bangun model JST
model = Sequential([
    # Ubah gambar 28x28 menjadi vektor
    Flatten(input_shape=(28, 28)),
    # Hidden layer 1
    Dense(128, activation='relu'),
    # Hidden layer 2
    Dense(64, activation='relu'),
    # Hidden layer 3
    Dense(32, activation='relu'),
    # Output layer (10 kelas)
    Dense(10, activation='softmax')
])
```

## Eksperimen 3. Bandingkan Sigmoid dan ReLU

```python
# 2. Bangun model JST
model = Sequential([
    # Ubah gambar 28x28 menjadi vektor
    Flatten(input_shape=(28, 28)),
    # Hidden layer 1
    Dense(128, activation='sigmoid'),
    # Hidden layer 2
    Dense(64, activation='sigmoid'),
    # Output layer (10 kelas)
    Dense(10, activation='softmax')
])
```

```python
# 3. Kompilasi model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
# 4. Latih model
mulai = time.perf_counter()
model.fit(X_train, y_train, epochs=10, batch_size=32)
waktu_latih = time.perf_counter() - mulai
```

```python
# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi pada data uji: {acc:.4f}")
hasil.append(('128-64, Sigmoid', loss, acc, waktu_latih))
```
